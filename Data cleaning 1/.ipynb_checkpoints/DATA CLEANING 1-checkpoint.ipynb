```python
import pandas as pd
df = pd.read_csv("dirty_cafe_sales.csv")
df.head()
```

```python
df.info()
```

```python
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce')
df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce')
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.drop(columns = ["Transaction ID"],inplace = True)
```

```python
df.head()
```

```python
df.Item.value_counts()
```

```python
df.Item.fillna("Juice",inplace= True)
```

```python
df.isnull().sum()
```

```python
df.Quantity.value_counts()
```

```python
df.Quantity.fillna(5,inplace= True)
df.isnull().sum()
```

```python
df["Price Per Unit"].value_counts()
```

```python
df['Price Per Unit'].fillna(3.0,inplace= True)
df.isnull().sum()
```

```python
df["Total Spent"].value_counts()
```

```python

df['Total Spent'].fillna(6.0,inplace= True)
df.isnull().sum()
```

```python
df["Payment Method"].value_counts()
```

```python
df = df[df['Payment Method'] != 'UNKNOWN']
df = df[df['Payment Method'] != 'ERROR']
df['Payment Method'].fillna("Digital Wallet",inplace= True)
df.isnull().sum()
```

```python
df["Location"].value_counts()
```

```python
df = df[df['Location'] != 'UNKNOWN']
df = df[df['Location'] != 'ERROR']
df['Location'].fillna("In-store",inplace= True)
df.isnull().sum()
```

```python
df["Transaction Date"].value_counts()
```

```python
df = df[df['Transaction Date'] != 'UNKNOWN']
df = df[df['Transaction Date'] != 'ERROR']
df['Transaction Date'].fillna("2023-02-06",inplace= True)
df.isnull().sum()
```

```python
df.head()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(ignore_index=True,inplace = True)
```
