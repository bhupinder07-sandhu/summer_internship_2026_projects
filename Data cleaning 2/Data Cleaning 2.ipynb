```python
df['Price Per Unit'].fillna(33.5,inplace =True)
df.isnull().sum()
```

```python
df['Quantity'].value_counts()
```

```python
df['Quantity'].fillna(10.0,inplace =True)
df.isnull().sum()
```

```python
df['Total Spent'].value_counts()
```

```python
df['Total Spent'].fillna(40.0,inplace =True)
df.isnull().sum()
```

```python
df['Discount Applied'].fillna('True',inplace =True)
df.isnull().sum()
```

```python
df.drop_duplicates(ignore_index=True,inplace =True)
```

```python
df.boxplot(column=['Price Per Unit','Quantity','Total Spent'])
plt.show()
```

```python
import numpy as np

# Select numerical columns
num_cols = df.select_dtypes(include=np.number).columns

# Remove outliers using IQR
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print("Shape after removing outliers:", df.shape)
```

```python
df.boxplot(column=['Price Per Unit','Quantity','Total Spent'])
plt.show()
```
