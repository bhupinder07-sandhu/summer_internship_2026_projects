```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

from sklearn.preprocessing import  LabelEncoder
from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
df = pd.read_csv('loan_approval_dataset.csv')
df.head()
```

```python
df.drop(columns ="loan_id",inplace = True)
df
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
import numpy as np
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print(df.shape)
```

```python
df.columns = df.columns.str.strip()

print(df.columns)
```

```python
le_education = LabelEncoder()
le_self_employed = LabelEncoder()
le_loan_status = LabelEncoder()
df['self_employed'] = le_self_employed.fit_transform(df['self_employed'])
df['education'] = le_education.fit_transform(df['education'])
df['loan_status'] = le_loan_status.fit_transform(df['loan_status'])

df
```

```python
df.head()
```

```python
x_loan_status = df.iloc[:,:-1]
print(x_loan_status)
y_loan_status = df.iloc[:,-1]
y_loan_status 
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(x_loan_status,y_loan_status,test_size=0.20,random_state=42)
X_train
```

```python
loan_status_model=RandomForestClassifier()
loan_status_model.fit(X_train, Y_train)
y_pred=loan_status_model.predict(X_test)
print(accuracy_score(Y_test,y_pred))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(Y_test, y_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(Y_test, y_pred))
```
